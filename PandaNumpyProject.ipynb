```python
import numpy as np
import pandas as pd
```

```python

# ================================
# STEP 1 — Data Setup
# ================================
marks = np.array([
    [85, 90, 78, 92],
    [45, 55, np.nan, 60],
    [90, 88, 95, 87],
    [30, np.nan, 40, 35],
    [70, 75, 80, np.nan],
    [55, 60, 58, 62]
])
students=["Ali","Sara","Fatima","Ayesha","Saira","Dua"]
subjects=["Java","Python","DBMS","C#"]
# ================================
# STEP 2 — NumPy Statistics
# ================================
print("="*60)
print("Overall class statistics (NUMPY)")
print("="*60)
print(f"Class mean    :{np.nanmean(marks):.2f}")
print(f"Highest marks    :{np.nanmax(marks):.0f}")
print(f"Lowest marks    :{np.nanmin(marks):.0f}")
print(f"Std Deviation  :{np.nanstd(marks):.2f} ")


# ================================
# STEP 3 — Create DataFrame
# ================================
df=pd.DataFrame(
    marks,
    index=students,
    columns=subjects
)
copyData=df.copy()


# ================================
# STEP 4 — Handle Missing Values
# ================================
copyData.fillna(copyData.mean(),inplace=True)

# ================================
# STEP 5 — Total, Average, Percentage
# ================================
copyData["Total"]=copyData[subjects].sum(axis=1).round(2)
copyData["Average"]=copyData[subjects].mean(axis=1).round(2)
total_marks=len(subjects)*100
copyData["Percentage"]=(copyData["Total"]/total_marks*100).round(2)

# ================================
# STEP 6 — Grade Function
# ================================

def assign_grade(avg):
    if avg>=90:
        return "A+"
    elif avg>=80:
        return "A"
    elif avg>=70:
        return "B"
    elif avg>=60:
        return "C"
    elif avg>50:
        return "D"
    else:
        return "F"

copyData["Grade"]=copyData["Average"].apply(assign_grade)

# ================================
# STEP 7 — Pass / Fail
# ================================

copyData["Result"]=np.where(
    copyData["Average"]>=50,
    "Pass",
    "Fail"
    )
# ================================
# STEP 8 — Ranking
# ================================
 
copyData["Rank"]=(
    copyData["Average"].rank(
        ascending=False,
        method="dense"
    ).
        astype(int)
    ) 


# ================================
# STEP 9 — Result Card
# ================================
print("\n")
print("=" * 110)
print("                         STUDENT RESULT CARD")
print("=" * 110)

print(
    f"{'Student':>10}"
    f"{'Java':>8}"
    f"{'Python':>10}"
    f"{'DBMS':>10}"
    f"{'C#':>8}"
    f"{'Total':>10}"
    f"{'Average':>10}"
    f"{'%age':>8}"
    f"{'Grade':>10}"
    f"{'Result':>8}"
    f"{'Rank':>10}"
)
print("-" * 110)
for student in students:
    row=copyData.loc[student]
    print(
        f"{student:<10}"
        f"{row['Java']:>8.0f}"
        f"{row['Python']:>10.0f}"
        f"{row['DBMS']:>10.0f}"
        f"{row['C#']:>8.0f}"
        f"{row['Total']:>10.0f}"
        f"{row['Average']:>10.2f}"
        f"{row['Percentage']:>8.2f}"
        f"{row['Grade']:>8}"
        f"{row['Result']:>10}"
        f"{row['Rank']:>8}"    
    )

# ================================
# STEP 10 — Class Summary
# ================================
print("\nCLASS SUMMARY")
print("=" * 60)

print(f"Total Students ::{len(copyData)}")
print(f"Passed          ::{(copyData['Result']=='Pass').sum()}")
print(f"Failed           ::{(copyData['Result']=='Fail').sum()}")
print(f"Class Average    ::{copyData['Average'].mean():.2f}  ")


top_student=copyData['Average'].idxmax()
top_score=copyData['Average'].max()
print(f"Top Student is ::{top_student}")
print(f"%age marks is  ::{top_score:.2f}")
# ================================
# STEP 11 — Subject Statistics
# ================================

print("\nSUBJECT-WISE STATISTICS")
print("=" * 60)

for subject in subjects:
    print(f"\n{subject}")
    print("-" * 20)
    print(f"Mean ::: {copyData[subject].mean():.2f}")
    print(f"Highest:::{copyData[subject].max()}")
    print(f"Lowest :::{copyData[subject].min()}")


# ================================
# STEP 12 — Top 3 Students
# ================================
print("\nTOP 3 STUDENTS")
print("=" * 60)

top3=copyData.sort_values(by="Average",ascending=False).head(3)
print(top3[["Average","Percentage","Grade","Rank"]])

# ================================
# STEP 13 — Grade Distribution
# ================================

print("\nGRADE DISTRIBUTION")
print("=" * 60)
print(copyData["Grade"].value_counts().sort_index())
# ================================
# STEP 14 — Save Results
# ================================
copyData.to_csv("student_result.csv")
copyData.to_excel("student_result.xlsx")
print("\nFile Saved Successfully:")
print("student_results.csv")

# ================================
# STEP 15 — Final DataFrame
# ================================
print("\nCOMPLETE DATAFRAME")
print("=" * 60)

print(copyData)
```
